Matrix Addition:

two matrices can be added only if they have the same shape

```python
#creating two matrices
a=np.array([[2,3],[10,5]])
b=np.array([[10,5],[20,4]])
```

```python
print(a)
a.shape
```

```python
print(b)
b.shape
```

Adding two matrices

```python
sum=a+b
print(sum)
sum.shape
```
