```python
import numpy as np
```

Creating a matrix using numpy

```python
matrix_1=np.array([[2,3],[6,7]])
print(matrix_1)
```

```python
matrix_1.shape
```

```python
matrix_2=np.array([[1,2,3],[1,2,3],[1,2,3]])
print(matrix_2)
```

```python
matrix_2.shape
```

creating a matrices with random values

```python
matrix_3=np.random.rand(3,3)
print(matrix_3)
```

```python
#creating a matrix with random integers
random_integer_matrix=np.random.randint(100,size=(4,5))
print(random_integer_matrix)
```

Matrix with all the values as one

```python
matrix_4=np.ones((2,3))  #floating values
print(matrix_4)
```

```python
matrix_4=np.ones((2,3),dtype=int)  #integer values
print(matrix_4)
```

Null matrix or Zero matrix

```python
matrix_5=np.zeros((2,3))  #floating values
print(matrix_5)
```

```python
matrix_5=np.zeros((2,3),dtype=int)  #integer values
print(matrix_5)
```

Identity Matrix

```python
identity_matrix=np.eye(3)  #floating values
print(identity_matrix)
```

```python
identity_matrix=np.eye(3,dtype=int)  #integer values
print(identity_matrix)
```

Transpose of a Matrix

```python
#matrix with random integer values
matrix=np.random.randint(100,size=(4,5))
print(matrix)
```

```python
transpose_matrix=np.transpose(matrix)
print(transpose_matrix)
```
